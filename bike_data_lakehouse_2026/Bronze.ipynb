# Bronze ingestion
Load CRM and ERP CSV files from the Unity Catalog volume into managed Delta tables.

## Load `cust_info.csv`

```sql
CREATE OR REPLACE TEMPORARY VIEW cust_info_csv
USING CSV
OPTIONS (
  path '/Volumes/workspace/bronze/source_systems/source_crm/cust_info.csv',
  header 'true',
  inferSchema 'true'
);

CREATE OR REPLACE TABLE workspace.bronze.crm_cust_info
USING DELTA
AS SELECT * FROM cust_info_csv;
```
